```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False

monthly = pd.read_csv('../data/monthly_gmv.csv')
city    = pd.read_csv('../data/city_gmv.csv')
cat     = pd.read_csv('../data/category_gmv.csv')
newc    = pd.read_csv('../data/new_user.csv')

# 月份转成日期
monthly['ym'] = pd.to_datetime(monthly['ym'].astype(str) + '-01')
newc['ym']    = pd.to_datetime(newc['ym'].astype(str) + '-01')

# 图1：月度 GMV 趋势
fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(monthly['ym'], monthly['gmv'], marker='o', linewidth=2)
ax.set_title('月度 GMV 趋势')
ax.set_ylabel('GMV（元）')
ax.grid(alpha=0.3)
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../images/01_月度GMV.png', dpi=150, bbox_inches='tight')
plt.show()

# 图2：城市 GMV 对比
c = city.sort_values('gmv')
fig, ax = plt.subplots(figsize=(8, 5))
ax.barh(c['city'], c['gmv'], color='#4C78A8')
ax.set_title('各城市 GMV')
ax.set_xlabel('GMV（元）')
plt.tight_layout()
plt.savefig('../images/02_城市GMV.png', dpi=150, bbox_inches='tight')
plt.show()

# 图3：品类占比
fig, ax = plt.subplots(figsize=(7, 7))
ax.pie(cat['gmv'], labels=cat['category'], autopct='%1.1f%%', startangle=90)
ax.set_title('各品类 GMV 占比')
plt.tight_layout()
plt.savefig('../images/03_品类占比.png', dpi=150, bbox_inches='tight')
plt.show()

# 图4：新客占比趋势
fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(newc['ym'], newc['new_ratio'], marker='o', color='#E45756')
ax.set_title('新客占比趋势')
ax.set_ylabel('新客占比（%）')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('../images/04_新客占比.png', dpi=150, bbox_inches='tight')
plt.show()
```
