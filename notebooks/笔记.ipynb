```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = {
    "id": [1,2,3,4,5,6,2],
    "name": ["A","B","C","D","E","F","B"],
    "class": ["1班","1班","1班","2班","2班","2班","1班"],
    "score": [88,92,76,92,85,60,92],
    "date": ["2025-01-01","2025-01-01","2025-01-02","2025-01-02","2025-01-03","2025-01-03","2025-01-01"]
}
df = pd.DataFrame(data)
print(df.shape)  # (7,5)
```

```python
print(df.shape) #写行和列，几行几列
```

```python
 print(df.head(1)) #查看前几行，其中默认为5行，如图1是第一行
```

```python
print(df.info()) #查看字段情况，尤其是损失
```

```python
print(df.describe())   #计算总数，平均数，标准差，还有各个百分位数
```

```python
print(df.isnull().sum()) #计算损失值的数量
```

```python
print(df["class"].value_counts()) #统计某个列每个值出现的次数
```

```python
print(df['name'].nunique())  #统计不同列中不同值出现的次数，注意会去重
```

```python
print(df['name'])  #查看单行
```

```python
print(df[['name' , 'class']])   #多行的操作
```

```python
print(df.loc[0,'name'])  #取出name列的第0行
```

```python
print(df.loc[0:2,['name', 'class']]) #取出name,class列的第0行到第2行
```

```python
print(df.iloc[0,1]) #取出第0行
```

```python
print(df.iloc[0:2,:]) #取出第0到第2行，然后取出对应的所有列
```

```python
print(df[df['score']>80]) #找出成绩大于80的信息
```

```python
print(df[(df['score']>80) & (df['class']=='1班')])  #查询成绩大于80和班级是一班的学生信息，注意不能用and，or
```

```python
print(df[(df['score']<=80) & (df['score']>=60)])
```
